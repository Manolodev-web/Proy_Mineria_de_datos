# Análisis Comparativo: XGBoost vs CatBoost

Este análisis evalúa el rendimiento de dos de las librerías más potentes de Gradient Boosting aplicadas a la producción de arroz. Para evidenciar sus diferencias estructurales, hemos incluido la variable categórica `DEPARTAMENTO`.

## 1. Tratamiento de Variables Categóricas (Eficiencia)
La principal diferencia técnica radica en cómo interpretan el texto:
* **XGBoost:** Requiere convertir el texto en columnas numéricas binarias (One-Hot Encoding). Esto provocó que nuestras 4 variables originales se multiplicaran, inflando la memoria y volviendo al modelo más propenso a ramificaciones ineficientes.
* **CatBoost:** Está diseñado para procesar el texto nativamente de forma interna (`cat_features`). Mantuvo la estructura original de 4 columnas, siendo estructuralmente más limpio y rápido.

## 2. Regresión (Predicción de Producción)
Ambos algoritmos lidian con los fuertes valores atípicos del dataset. Sin embargo, su arquitectura interna provoca distintos resultados en la distribución de sus residuos:
* A pesar del escalamiento y la regularización que ambos aplican, las curvas muestran la dificultad inherente de predecir la varianza extrema de los grandes productores, manteniendo un R² conservador pero honesto.

## 3. Clasificación (Detección de Invierno con Balanceo)
Para combatir el desbalance (1235 Verano vs 140 Invierno):
* En **XGBoost** tuvimos que calcular y asignar manualmente el ratio de desbalance mediante el parámetro `scale_pos_weight`.
* En **CatBoost**, bastó con activar `auto_class_weights='Balanced'`.

Ambos modelos lograron forzar el aprendizaje de la clase minoritaria, lo cual se refleja en las matrices de confusión, donde lograron identificar correctamente una porción significativa de la campaña de Invierno, a cambio de clasificar algunos falsos positivos en Verano (un compromiso estadístico necesario para el balanceo).
