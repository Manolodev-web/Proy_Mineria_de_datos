# Análisis de Clasificación: KNN y Regresión Logística

Este cuaderno documenta el análisis de los modelos de clasificación aplicados al dataset de producción de arroz, con el objetivo de predecir la **Campaña Agrícola (Verano vs Invierno)**.

## 1. K-Nearest Neighbors (KNN)

**Métricas obtenidas en consola:**
* **Mejor K encontrado:** 19
* **Precisión en Prueba (Accuracy):** 89.35%

**Análisis de la Matriz de Confusión:**
A pesar de la alta precisión global (~89%), la matriz de confusión revela un sesgo masivo. El modelo KNN predijo correctamente casi todos los casos de Verano (368 aciertos, 3 errores)[cite: 7], pero falló en detectar casi por completo la campaña de Invierno (1 acierto, 41 errores)[cite: 7]. Esto evidencia la "Paradoja de la Precisión" debido al fuerte desbalance de clases en los datos originales (1235 Verano vs 140 Invierno). Al basarse en la mayoría de los "vecinos" (K=19), el modelo casi siempre predecirá "Verano".

**Análisis de Fronteras de Decisión:**
La gráfica de fronteras ilustra espacialmente este sesgo. La zona roja (decisión para Verano) domina la totalidad del gráfico[cite: 8]. Los escasos puntos verdes (Invierno) están fuertemente solapados con los rojos en la esquina inferior izquierda, impidiendo que el algoritmo defina un límite claro entre ambas campañas utilizando únicamente la Superficie y la Producción[cite: 8].

## 2. Regresión Logística

**Métricas obtenidas en consola:**
* **Precisión Global (Accuracy):** 90.55%

**Análisis de la Matriz de Confusión:**
El modelo multivariado de Regresión Logística sufre exactamente el mismo problema. Identificó perfectamente 247 casos de Verano (0 falsos positivos)[cite: 9], pero solo fue capaz de detectar 2 de los 28 casos reales de Invierno[cite: 9].

<img src="../03_imagenes/05_logistica_matriz.png" width="600" />

**Análisis de la Curva Logística (Modelo Univariado):**
Al graficar la probabilidad de que un cultivo sea de Invierno basada solo en el "Rendimiento (Kg/Ha)", vemos el impacto de los valores atípicos. La inmensa mayoría de los productores (azules y rojos) están concentrados en rendimientos bajos (cerca del 0 en el eje X)[cite: 10]. Sin embargo, la presencia de un único cultivo de invierno con un rendimiento anómalo cercano a los 70,000 Kg/Ha estira toda la curva logística hacia la derecha[cite: 10]. Como resultado, para el 99% de los agricultores comunes, el modelo calcula una probabilidad de ser "Invierno" por debajo del umbral del 50%, clasificándolos erróneamente como "Verano"[cite: 10].

### Conclusión Técnica
Ambos modelos estándar fracasan en su objetivo principal debido al desbalance de clases y a la presencia de valores extremos (outliers). Para problemas médicos, financieros o agrícolas donde detectar la clase minoritaria es vital, es imperativo utilizar algoritmos que permitan ponderar pesos (como se verá en los ensambles) o aplicar técnicas de sub/sobremuestreo de datos (SMOTE) antes del entrenamiento.
