# Análisis de Máquinas de Vectores de Soporte (SVM)

Este cuaderno evalúa la aplicación de modelos SVM, los cuales proyectan los datos escalados a múltiples dimensiones para trazar el "margen más ancho posible".

## 1. SVR: Regresión de Producción de Arroz

**Métricas obtenidas en consola:**
* **R² Score:** -0.0080
* **MAE:** 4226.07 qq
* **RMSE:** 40355.34 qq

**Análisis del comportamiento:**
La variante de regresión (SVR) busca crear un "tubo" con un margen de tolerancia donde encajen los puntos. En nuestro dataset, los valores atípicos son tan extremos que SVR fracasa por completo, arrojando un R² negativo (-0.0080). Un R² negativo indica que el modelo rinde *peor* que si simplemente adivinara el valor promedio en cada caso. La gráfica demuestra cómo la línea de "predicción perfecta" se aleja totalmente de la realidad modelada por el SVR para el mega-productor[cite: 27].

## 2. SVC: Clasificación de Campañas Agrícolas

**Métricas obtenidas en consola:**
* **Accuracy General:** 83.05%

**Análisis de la Matriz de Confusión:**
A diferencia de la regresión, el clasificador (SVC) con el parámetro `class_weight='balanced'` logra converger de forma útil.
* Logró identificar correctamente 16 de los 42 casos reales de Invierno (un *recall* del 38%).
* A cambio, generó 44 falsos positivos (clasificando Verano como Invierno)[cite: 28].
Aunque estas métricas de sensibilidad son inferiores a las logradas por CatBoost (que detectó 35 casos), SVC demuestra capacidad operativa frente al desbalance[cite: 28].

**Análisis de Fronteras de Decisión y Vectores de Soporte:**
Al comprimir las variables predictoras en 2D mediante PCA, visualizamos cómo trabaja el kernel **RBF (Radial Basis Function)**. A diferencia de las líneas rectas de los árboles, RBF traza "islas" curvas y complejas[cite: 29]. 
La zona azul (Verano) intenta agrupar la densa nube de puntos, mientras que la zona roja (Invierno) abarca el resto del plano[cite: 29]. Los anillos amarillos resaltan los **Vectores de Soporte**, que son los puntos de datos cruciales que el algoritmo utilizó matemáticamente para anclar y dar forma a esas fronteras sinuosas[cite: 29].
