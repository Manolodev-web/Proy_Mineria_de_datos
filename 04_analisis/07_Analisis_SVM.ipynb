# Análisis de Máquinas de Vectores de Soporte (SVM)

SVM es un modelo geométrico que busca proyectar los datos al espacio multidimensional para trazar el "margen más ancho posible" que separe las clases o contenga las predicciones. Este enfoque requiere estrictamente que los datos estén normalizados (StandardScaler).

## 1. SVR: Regresión de Producción de Arroz
La variante de regresión (Support Vector Regression) busca crear un "tubo" con un margen de tolerancia (épsilon) donde encajen la mayor cantidad de puntos posibles. 
En nuestra base de datos, los valores de producción son tan extremos y dispersos (con el 95% agrupado cerca del origen y el 5% disparado a valores inmensos) que SVR fracasa matemáticamente, arrojando un R² negativo (aprox. -0.008). Esto significa que intentar predecir usando SVR con un kernel RBF bajo esta escala es menos preciso que si simplemente predijéramos la media en todos los casos.

## 2. SVC: Clasificación de Campañas Agrícolas
A diferencia de la regresión, el clasificador (Support Vector Classifier) sí logra converger utilizando el parámetro `class_weight='balanced'`. 
* **Vectores de Soporte y PCA:** Dado que tenemos 3 dimensiones predictoras, utilizamos Análisis de Componentes Principales (PCA) para comprimirlas en 2D y poder visualizar la frontera matemática. 
* El kernel **RBF (Radial Basis Function)** permite que el modelo trace curvas complejas en lugar de líneas rectas para aislar a la clase minoritaria (Invierno, en rojo) de la masa dominante (Verano, en azul). Los puntos envueltos en círculos amarillos son los vectores de soporte: los puntos críticos de datos que sostienen la estructura de la frontera.
* La Matriz de Confusión confirma que la estructura curva es capaz de detectar a la clase minoritaria con una precisión razonable, manteniendo un equilibrio funcional a pesar del desbalance original.
