# Análisis Comparativo: Ensambles de Boosting (AdaBoost vs LightGBM)

Este cuaderno evalúa la implementación de técnicas de potenciación (Boosting), donde múltiples modelos (usualmente árboles) se entrenan de forma secuencial para corregir los errores de sus predecesores. 

## 1. Regresión (Predicción de Producción)

**Métricas obtenidas en consola:**
* **AdaBoost:** R² = 0.3761 | MAE = 2979.76 qq
* **LightGBM:** R² = 0.2154 | MAE = 2559.12 qq

**Análisis del comportamiento:**
Estamos ante una clásica "paradoja de métricas" provocada por valores atípicos. LightGBM presenta un Error Absoluto Medio (MAE) inferior al de AdaBoost (2559 vs 2979), lo que indica que para la gran mayoría de los pequeños productores, LightGBM es más preciso. La gráfica de distribución de residuos confirma esto: la campana de LightGBM (verde) es muchísimo más estrecha y concentrada en el cero que la de AdaBoost (naranja)[cite: 20]. 

Sin embargo, LightGBM tiene un R² mucho peor (0.21 vs 0.37). Al observar la gráfica de dispersión, descubrimos la razón: ante el productor extremo de 800,000 qq, AdaBoost predice un valor cercano a los 150,000 qq, mientras que LightGBM es aún más conservador, prediciendo menos de 100,000 qq[cite: 19]. Dado que el R² penaliza severamente los errores grandes al elevarlos al cuadrado, el "fallo" de LightGBM en este único punto colapsa su métrica general, a pesar de ser un modelo más ajustado para la mayoría poblacional.

## 2. Clasificación (Detección de la Campaña de Invierno)

**Métricas obtenidas en consola:**
* **AdaBoost - Accuracy:** 89.83%
* **LightGBM - Accuracy:** 83.54%

**Análisis de Matrices de Confusión y Balanceo:**
La aparente superioridad del Accuracy de AdaBoost (89.83%) es un espejismo estadístico. Dado que el 89.8% de nuestro set de pruebas (371 de 413) pertenece a la campaña de Verano, AdaBoost alcanza esta métrica simplemente prediciendo "Verano" casi todo el tiempo. 

LightGBM, al haber sido entrenado con el parámetro de balanceo de clases, sacrifica intencionalmente su precisión global (bajando al 83.54%) para forzarse a aprender los patrones de la clase minoritaria. La matriz de confusión de LightGBM demuestra que logró detectar correctamente 11 casos de Invierno, a cambio de cometer 37 falsos positivos (clasificando Verano como Invierno)[cite: 21]. 

**Análisis de Curvas ROC:**
Las curvas ROC confirman que las capacidades matemáticas subyacentes de ambos modelos son muy similares. AdaBoost logra un Área Bajo la Curva (AUC) de 0.710, ligeramente superior al 0.696 de LightGBM[cite: 22]. Estas curvas escalonadas demuestran que, independientemente del umbral de probabilidad elegido, predecir la estacionalidad utilizando solo la Superficie y el Rendimiento sigue siendo un desafío complejo con un límite claro de separabilidad[cite: 22].
