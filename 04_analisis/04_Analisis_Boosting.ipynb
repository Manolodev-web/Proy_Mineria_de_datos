# Análisis Comparativo: Ensambles de Boosting (AdaBoost vs LightGBM)

Este cuaderno evalúa la implementación de técnicas de potenciación (Boosting), las cuales entrenan múltiples modelos secuenciales donde cada nuevo modelo intenta corregir los errores de su predecesor.

## 1. Regresión (Predicción de Producción)
A diferencia de los modelos polinomiales, los ensambles basados en árboles (como AdaBoost y LightGBM) no se desbordan intentando alcanzar el valor atípico gigante, sino que mantienen predicciones mucho más conservadoras y realistas para la mayoría de los productores.
* **AdaBoost** tiende a verse más afectado por el outlier extremo, ya que su naturaleza es precisamente darle más "peso" a los datos donde el error es grande, lo que puede causar que el modelo se desestabilice intentando predecir ese punto anómalo.
* **LightGBM**, al estar basado en el crecimiento de hojas (leaf-wise) y el agrupamiento de gradientes, gestiona mejor esta asimetría, reflejando una densidad de residuos mucho más compacta.

## 2. Clasificación (Detección de la Campaña de Invierno)
El reto principal en este dataset es el fuerte desbalance de clases (Verano vs Invierno).
* **AdaBoost:** Por defecto, prioriza la precisión general (Accuracy). Al no contar con un hiperparámetro nativo de balanceo de pesos, la matriz de confusión demuestra que sigue ignorando a la clase minoritaria (Invierno) casi por completo.
* **LightGBM:** Su implementación nos permite activar el parámetro `class_weight='balanced'`. La matriz de confusión resultante muestra un comportamiento diametralmente opuesto: sacrifica ligeramente el acierto en la campaña de verano para lograr detectar un gran volumen de la campaña de invierno.
* **Curva ROC:** Al observar el Área Bajo la Curva (AUC), ambos modelos demuestran tener la capacidad matemática de separar las clases (con un AUC > 0.85), pero LightGBM se vuelve operativamente superior gracias al manejo interno de los pesos.
