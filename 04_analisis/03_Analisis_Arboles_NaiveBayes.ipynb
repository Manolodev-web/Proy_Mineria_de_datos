# Análisis Comparativo: Árboles de Decisión y Naive Bayes

Este cuaderno documenta el rendimiento de dos algoritmos (partición geométrica vs. probabilidad condicional) aplicados a los problemas de regresión y clasificación del dataset de arroz.

## 1. Árbol de Decisión: Regresión (Producción)

**Métricas obtenidas:**
* **R² Test:** 0.3726
* **MAE:** 2660.00 qq
* **RMSE:** 31837.48 qq
* **MAPE:** 66.10 %

**Análisis del comportamiento:**
El Árbol de Decisión arroja un R² de 37.26%, un valor inferior a los modelos polinomiales, pero su comportamiento es resistente a los valores atípicos. Al observar la gráfica de dispersión, el modelo no intenta trazar una curva que alcance al productor de 800,000 qq; en su lugar, agrupa a los productores masivos en un nodo "tope", prediciendo un valor constante para los extremos[cite: 13, 14]. 
El algoritmo realiza su primer corte principal en `SUPERFICIE (Ha) <= 854.8`[cite: 13]. Luego, aísla los valores gigantes en nodos específicos (como el de 169,066 qq), evitando que ese dato distorsione las predicciones de los miles de pequeños productores ubicados en la base[cite: 13].

## 2. Naive Bayes: Regresión (Discretizada)

**Métricas obtenidas:**
* **Accuracy Global:** 50.12%

**Análisis del comportamiento:**
Para forzar a Naive Bayes a predecir producción, discretizamos la variable continua en tres rangos (Baja, Media, Alta). El modelo acierta en la mitad de los casos (50.12%), pero la matriz revela fallos estructurales.
El modelo es incapaz de identificar la producción "Baja": de los 141 casos reales de producción baja, clasificó erróneamente 118 como "Media" y solo acertó 23[cite: 15]. Se desempeña marginalmente mejor en la categoría "Alta", donde logra 76 aciertos frente a 60 confusiones con la clase "Media"[cite: 15]. Asumir independencia entre la Superficie y el Rendimiento para clasificar densidades de producción no es un enfoque óptimo.

## 3. Árbol de Decisión: Clasificación (Campaña Agrícola)

**Métricas obtenidas:**
* **Accuracy Global:** 80.15%

**Análisis de la Matriz de Confusión y Balanceo:**
Implementamos el hiperparámetro `class_weight='balanced'`. El modelo logró identificar correctamente 24 de los 42 casos reales de Invierno en el set de prueba[cite: 16]. Para lograr este salto en sensibilidad, el modelo clasificó a 64 productores de Verano como si fueran de Invierno (falsos positivos)[cite: 16].

**Análisis de la Estructura:**
El nodo raíz discrimina inmediatamente por `SUPERFICIE (Ha) <= 39.9`[cite: 17]. A partir de ahí, el algoritmo utiliza el `RENDIMIENTO (Kg/Ha)` para encontrar "bolsas" de datos donde la campaña de invierno es dominante (nodos azules), demostrando que la combinación de áreas de cultivo con ciertos rendimientos define la temporada[cite: 17].

## 4. Naive Bayes: Clasificación (Multinomial)

**Métricas obtenidas:**
* **Accuracy Global:** 88.62%

**Análisis de la Matriz de Confusión:**
Utilizando las variables geográficas categóricas bajo el enfoque Multinomial, Naive Bayes identificó correctamente 356 casos de Verano con 15 errores[cite: 18]. En la campaña de Invierno, solo logró clasificar correctamente 10 casos, fallando en 32[cite: 18]. Aunque su rendimiento en la clase minoritaria es ligeramente superior al de un KNN básico, queda por detrás del Árbol de Decisión balanceado, demostrando que la ubicación geográfica por sí sola no es suficiente para predecir la estacionalidad[cite: 18].
