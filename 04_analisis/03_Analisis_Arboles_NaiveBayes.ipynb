# Análisis Comparativo: Árboles de Decisión y Naive Bayes

Este análisis evalúa el comportamiento de dos enfoques algorítmicos distintos aplicados tanto a tareas de **Regresión** como de **Clasificación** sobre el dataset de producción de arroz.

## 1. Modelos de Regresión (Predicción de Producción en qq)
* **Árbol de Decisión (Regressor):** Al segmentar jerárquicamente las hectáreas y el rendimiento, maneja de forma eficiente la asimetría de los datos sin verse arrastrado desproporcionadamente por los valores extremos individuales.
<img src="../03_imagenes/07_dt_regresion_arbol.png" width="700" />
<img src="../03_imagenes/08_dt_regresion_scatter.png" width="600" />

* **Naive Bayes (Adaptado por Bins):** Al requerir una conversión previa de la variable continua en rangos discretos (Baja, Media, Alta), el modelo evalúa probabilidades condicionales, mostrando cómo se distribuyen las densidades de producción según el espacio muestral.
<img src="../03_imagenes/09_nb_regresion_cm.png" width="600" />

---

## 2. Modelos de Clasificación (Predicción de Campaña Agrícola)
* **Árbol de Decisión (Classifier):** Gracias a la utilización del parámetro de balanceo (`class_weight='balanced'`), el modelo compensa la escasez de registros de la campaña de invierno frente a la de verano, generando particiones lógicas y transparentes interpretables a través de su estructura de nodos.
<img src="../03_imagenes/10_dt_clasificacion_cm.png" width="600" />
<img src="../03_imagenes/11_dt_clasificacion_arbol.png" width="700" />

* **Naive Bayes (Multinomial Classifier):** Utilizando las variables categóricas geográficas (Departamento y Provincia) bajo el supuesto de independencia condicional, el algoritmo clasifica con rapidez las probabilidades de pertenencia a cada campaña.
